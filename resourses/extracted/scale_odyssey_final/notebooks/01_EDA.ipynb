# SCALE x ODYSSEY -- 01: Exploratory Data Analysis

Visualize dataset distribution, sample images, and class balance.

```python
import sys
sys.path.insert(0, '../src')

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from dataset import AstroDataset, CLASS_NAMES_DISPLAY

train_ds = AstroDataset('../data/processed', 'train', 224)
val_ds = AstroDataset('../data/processed', 'val', 224)
test_ds = AstroDataset('../data/processed', 'test', 224)
print(f'Train: {len(train_ds)} | Val: {len(val_ds)} | Test: {len(test_ds)}')
```

## Class Distribution

```python
from collections import Counter

def plot_distribution(labels, title):
    counts = Counter(labels)
    plt.figure(figsize=(10, 5))
    plt.bar(range(5), [counts[i] for i in range(5)], color=['#3498db','#2ecc71','#9b59b6','#e67e22','#e74c3c'])
    plt.xticks(range(5), CLASS_NAMES_DISPLAY, rotation=30, ha='right')
    plt.title(f'{title} Class Distribution')
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()

plot_distribution(train_ds.labels, 'Train')
plot_distribution(val_ds.labels, 'Validation')
plot_distribution(test_ds.labels, 'Test')
```

## Sample Images (1 per class)

```python
fig, axes = plt.subplots(1, 5, figsize=(20, 4))
for i in range(5):
    idx = train_ds.labels.index(i)
    img = Image.open(train_ds.samples[idx])
    axes[i].imshow(img)
    axes[i].set_title(CLASS_NAMES_DISPLAY[i], fontsize=11)
    axes[i].axis('off')
plt.suptitle('Sample Images (1 per class)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Image Size Statistics

```python
widths, heights = [], []
for path in train_ds.samples[:500]:
    w, h = Image.open(path).size
    widths.append(w)
    heights.append(h)
print(f'Width  -- Mean: {np.mean(widths):.0f}, Min: {min(widths)}, Max: {max(widths)}')
print(f'Height -- Mean: {np.mean(heights):.0f}, Min: {min(heights)}, Max: {max(heights)}')
```
